# Transcrevendo uma imagem em texto com AWS Textract

Este notebook envia uma imagem com uma lista de material escolar ao Amazon Textract e imprime as linhas de texto identificadas.

```python
import boto3
from pathlib import Path

IMAGE_PATH = Path('lista_material_escolar.jpg')
REGION = 'us-east-1'
```

```python
client = boto3.client('textract', region_name=REGION)

with IMAGE_PATH.open('rb') as image_file:
    image_bytes = image_file.read()

response = client.detect_document_text(
    Document={'Bytes': image_bytes}
)
```

```python
print('=== TEXTO EXTRAÍDO ===')

for block in response['Blocks']:
    if block['BlockType'] == 'LINE':
        print(block['Text'])
```
